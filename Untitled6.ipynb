```python
import io
import re
import threading
from datetime import datetime
from urllib.parse import urlparse

import gradio as gr
import matplotlib.pyplot as plt
import pandas as pd
import qrcode

from IPython.display import display
```

```python
EVENTS = []
EVENT_LOCK = threading.Lock()


def log_event(event_type):
    event = {
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "event_type": event_type,
    }

    with EVENT_LOCK:
        EVENTS.append(event)

    return event


def register_visit():
    log_event("qr_page_visit")

    return (
        "Visit recorded anonymously. "
        "This is an authorized cybersecurity awareness simulation."
    )


def register_demo_submission():
    log_event("demo_interaction")

    return (
        "Simulation complete. No passwords or personal information "
        "were collected. Always inspect QR destinations before interacting."
    )
```

```python
SHORTENER_DOMAINS = {
    "bit.ly",
    "tinyurl.com",
    "t.co",
    "cutt.ly",
    "shorturl.at",
    "rb.gy",
}


def analyze_url(url):
    url = url.strip()

    if not url:
        return "Please enter a URL."

    if not url.startswith(("http://", "https://")):
        url = "https://" + url

    parsed = urlparse(url)
    hostname = parsed.hostname or ""

    risk_score = 0
    indicators = []

    if parsed.scheme != "https":
        risk_score += 25
        indicators.append("Connection does not use HTTPS.")

    if "@" in parsed.netloc:
        risk_score += 30
        indicators.append("URL contains an @ symbol, which can obscure the destination.")

    if hostname.startswith("xn--") or ".xn--" in hostname:
        risk_score += 25
        indicators.append("Domain uses punycode and may imitate another website.")

    if re.fullmatch(r"\d{1,3}(?:\.\d{1,3}){3}", hostname):
        risk_score += 25
        indicators.append("Destination uses an IP address instead of a domain name.")

    if hostname in SHORTENER_DOMAINS:
        risk_score += 20
        indicators.append("Shortened URL conceals its final destination.")

    if len(url) > 100:
        risk_score += 10
        indicators.append("Unusually long URL.")

    suspicious_terms = [
        "verify-account",
        "reset-password",
        "urgent-login",
        "confirm-identity",
        "security-update",
    ]

    if any(term in url.lower() for term in suspicious_terms):
        risk_score += 20
        indicators.append("URL contains wording commonly used in phishing messages.")

    risk_score = min(risk_score, 100)

    if risk_score >= 60:
        classification = "HIGH RISK"
    elif risk_score >= 30:
        classification = "MEDIUM RISK"
    else:
        classification = "LOW RISK"

    if not indicators:
        indicators.append(
            "No obvious indicators detected. This does not guarantee the URL is safe."
        )

    findings = "\n".join(f"- {indicator}" for indicator in indicators)

    return f"""
Classification: {classification}
Risk score: {risk_score}/100
Domain: {hostname}

Indicators:
{findings}
"""
```

```python
def build_dashboard():
    with EVENT_LOCK:
        events = list(EVENTS)

    if not events:
        figure, axis = plt.subplots(figsize=(7, 4))
        axis.text(
            0.5,
            0.5,
            "No awareness events recorded yet.",
            ha="center",
            va="center",
        )
        axis.axis("off")

        return "No events recorded yet.", figure

    dataframe = pd.DataFrame(events)

    visits = int((dataframe["event_type"] == "qr_page_visit").sum())
    interactions = int(
        (dataframe["event_type"] == "demo_interaction").sum()
    )

    interaction_rate = (
        round((interactions / visits) * 100, 2)
        if visits
        else 0
    )

    summary = f"""
QR landing-page visits: {visits}
Demo interactions: {interactions}
Interaction rate: {interaction_rate}%

Only anonymous event types and timestamps are recorded.
"""

    counts = dataframe["event_type"].value_counts()

    figure, axis = plt.subplots(figsize=(8, 4))

    counts.plot(
        kind="bar",
        ax=axis,
        color=["#22c55e", "#f59e0b"],
    )

    axis.set_title("QShield Awareness Campaign Events")
    axis.set_xlabel("Event type")
    axis.set_ylabel("Count")

    plt.xticks(rotation=15)
    plt.tight_layout()

    return summary, figure


def export_report():
    with EVENT_LOCK:
        events = list(EVENTS)

    dataframe = pd.DataFrame(
        events,
        columns=["timestamp", "event_type"],
    )

    output_path = "/content/qshield_awareness_report.csv"

    dataframe.to_csv(output_path, index=False)

    return output_path
```

```python
with gr.Blocks(title="QShield") as app:

    gr.Markdown(
        """
        # QShield

        ### QR Phishing Awareness Laboratory

        Authorized educational simulation for understanding suspicious QR
        destinations and measuring anonymous awareness interactions.

        **Never enter real passwords, financial details, or personal information.**
        """
    )

    visit_notice = gr.Textbox(
        label="Session notice",
        interactive=False,
    )

    app.load(
        fn=register_visit,
        outputs=visit_notice,
    )

    with gr.Tab("Awareness Simulation"):

        gr.Markdown(
            """
            ## Clearly Labeled Demonstration

            This page simulates the type of interaction a suspicious QR code
            might lead to.

            All displayed information is fictional. This interface does not
            request or store real usernames, passwords, or personal details.
            """
        )

        gr.Textbox(
            label="Fictional demo account",
            value="demo@example.com",
            interactive=False,
        )

        gr.Dropdown(
            choices=["DEMO-123", "TRAINING-456"],
            value="DEMO-123",
            label="Fictional training code",
        )

        demo_button = gr.Button(
            "Complete Safe Simulation",
            variant="primary",
        )

        demo_result = gr.Textbox(
            label="Awareness feedback",
            interactive=False,
        )

        demo_button.click(
            fn=register_demo_submission,
            outputs=demo_result,
        )

    with gr.Tab("URL Safety Analyzer"):

        url_input = gr.Textbox(
            label="Enter a URL",
            placeholder="https://example.com",
        )

        analyze_button = gr.Button("Analyze URL")

        url_result = gr.Textbox(
            label="Risk assessment",
            lines=10,
            interactive=False,
        )

        analyze_button.click(
            fn=analyze_url,
            inputs=url_input,
            outputs=url_result,
        )

    with gr.Tab("Analytics Dashboard"):

        summary = gr.Textbox(
            label="Campaign summary",
            lines=7,
            interactive=False,
        )

        chart = gr.Plot(label="Anonymous event analytics")

        refresh_button = gr.Button("Refresh Dashboard")

        refresh_button.click(
            fn=build_dashboard,
            outputs=[summary, chart],
        )

        export_button = gr.Button("Export CSV Report")

        downloadable_file = gr.File(
            label="Download anonymous campaign data",
        )

        export_button.click(
            fn=export_report,
            outputs=downloadable_file,
        )
```

```python
launch_result = app.launch(
    share=True,
    prevent_thread_lock=True,
)

public_url = getattr(app, "share_url", None)

if not public_url and len(launch_result) >= 3:
    public_url = launch_result[2]

print("QShield public URL:")
print(public_url)
```

```python
if not public_url:
    raise RuntimeError(
        "No public URL was generated. Check the previous cell."
    )

qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=4,
)

qr.add_data(public_url)
qr.make(fit=True)

qr_image = qr.make_image(
    fill_color="#101820",
    back_color="white",
)

qr_path = "/content/qshield_awareness_qr.png"

qr_image.save(qr_path)

display(qr_image)

print("QR saved at:", qr_path)
print("Destination:", public_url)
```
